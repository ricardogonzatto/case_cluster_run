```python
LAx_dash_Lcircle = [0.9, 1.0, 1.1, 1.2, 1.3]
```

```python
from matplotlib.ticker import LogLocator, LogFormatterMathtext, NullFormatter

plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "cm"

def plot_kl(kl_tilted, kl_untilted, title=None, ylim=(0.4, 300)):
    fig, ax = plt.subplots(figsize=(8, 3))
    ax.plot(LAx_dash_Lcircle, kl_tilted, "o-", color="tab:red", alpha=0.5, markersize=8,
            label=r"$\boldsymbol{x}_0 = (-0.1, 0.34, 0)$")
    ax.plot(LAx_dash_Lcircle, kl_untilted, "o-", color="tab:blue", alpha=0.5, markersize=8,
            label=r"$\boldsymbol{x}_0 = (0, 0, 0)$")
    ax.set_yscale("log")
    

    # y axis: labels only at powers of ten (10^n)
    all_y = np.concatenate([kl_tilted, kl_untilted])
    ax.set_ylim(10**np.floor(np.log10(all_y.min())), 10**np.ceil(np.log10(all_y.max())))
    ax.yaxis.set_major_locator(LogLocator(base=10))
    ax.yaxis.set_major_formatter(LogFormatterMathtext(base=10))
    ax.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10)))
    ax.yaxis.set_minor_formatter(NullFormatter())
    ax.set_ylim(*ylim)
    # x axis: only the values in LAx_dash_Lcircle
    ax.set_xticks(LAx_dash_Lcircle)
    ax.set_xticklabels([str(x) for x in LAx_dash_Lcircle])

    # vertical reference line at LAx/Lcircle = 1
    ax.axvline(1.0, color="k", linestyle="--", linewidth=1.5)

    # horizontal reference line at LAx/Lcircle = 1
    ax.axhline(1.0, color="k", linestyle="--", linewidth=1.5)

    # dashed grid on every tick (x ticks and all y ticks)
    ax.grid(True, which="both", linestyle="--", linewidth=0.6, alpha=0.5, color="gray")

    ax.set_xlabel(r"$L_{A_x}/L_{\rm circle}$", fontsize=18)
    ax.set_ylabel(r"$D_{\rm KL}(p\|q)$", fontsize=18)
    if title:
        ax.set_title(title)
    ax.legend(loc="upper right")
    fig.tight_layout()
    plt.show()

plot_kl(T_kl_E7_tilt, T_kl_E7_unt, "Tensor Modes - E7: Only TT")
plot_kl(T_kl_E9_tilt, T_kl_E9_unt, "Tensor Modes - E9: Only TT")
```

```python

def kl_panel_image(kl_tilted, kl_untilted, ylim=(0.4, 300), size_px=(2690, 1028), dpi=300):
    """Draw the KL plot (same style as plot_kl) and return it as an RGBA image
    with the same pixel size as the paper2b PNGs, so both columns match."""
    style = {"font.size": 14, "lines.linewidth": 2.2, "lines.markersize": 10,
             "axes.linewidth": 1.4, "legend.fontsize": 15}
    with plt.rc_context(style):
        fig, ax = plt.subplots(figsize=(size_px[0] / dpi, size_px[1] / dpi), dpi=dpi)
        _draw_kl(ax, kl_tilted, kl_untilted, ylim)
        fig.tight_layout()
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=dpi)
    plt.close(fig)
    buf.seek(0)
    return plt.imread(buf)


def _draw_kl(ax, kl_tilted, kl_untilted, ylim):
    ax.plot(LAx_dash_Lcircle, kl_tilted, "o-", color="tab:red", alpha=0.5,
            label=r"$\boldsymbol{x}_0 = (-0.1, 0.34, 0)$")
    ax.plot(LAx_dash_Lcircle, kl_untilted, "o-", color="tab:blue", alpha=0.5,
            label=r"$\boldsymbol{x}_0 = (0, 0, 0)$")
    ax.set_yscale("log")
    ax.yaxis.set_major_locator(LogLocator(base=10))
    ax.yaxis.set_major_formatter(LogFormatterMathtext(base=10))
    ax.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10)))
    ax.yaxis.set_minor_formatter(NullFormatter())
    ax.set_ylim(*ylim)
    ax.set_xticks(LAx_dash_Lcircle)
    ax.set_xticklabels([str(x) for x in LAx_dash_Lcircle])
    ax.axvline(1.0, color="k", linestyle="--", linewidth=2.2)
    ax.axhline(1.0, color="k", linestyle="--", linewidth=2.2)
    ax.grid(True, which="both", linestyle="--", linewidth=0.6, alpha=0.5, color="gray")
    ax.set_xlabel(r"$L_{A_x}/L_{\rm circle}$", fontsize=18)
    ax.set_ylabel(r"$D_{\rm KL}(p\|q)$", fontsize=18)
    ax.legend(loc="upper right")

# rows: E7, E9 | columns: tensor modes (TT), scalar paper 2b
# (argument order kept as in the cell above: plot_kl(T_kl_unt, T_kl_tilt))
rows = [
    ("E7", kl_panel_image(T_kl_E7_tilt, T_kl_E7_unt), plt.imread("E7_scalar_paper2b.png")),
    ("E9", kl_panel_image(T_kl_E9_tilt, T_kl_E9_unt), plt.imread("E9_scalar_paper2b.png")),
]

fig, axes = plt.subplots(2, 2, figsize=(16, 6.6))
for (name, img_tensor, img_paper), (ax_t, ax_p) in zip(rows, axes):
    ax_t.imshow(img_tensor)
    ax_p.imshow(img_paper)
    ax_t.set_title(f"Tensor Modes - {name}: Only TT", fontsize=14)
    ax_p.set_title(f"Scalar (paper 2b) - {name}", fontsize=14)
    for ax in (ax_t, ax_p):
        ax.axis("off")

# common manifold parameters shared by all four panels
manifold_label = (r"$L_{A_y} = 0,\ L_y = 1.4,\ L_{2x} = 0.7,\ L_{2z} = 1$")
fig.supxlabel(manifold_label, fontsize=18, y=0.0)

fig.tight_layout()
fig.savefig("grid_E7_E9_tensor_vs_paper2b.png", dpi=400, bbox_inches="tight")
plt.show()
```
